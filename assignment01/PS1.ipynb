## 2. Matrix multiplication

```python
import numpy as np

M1 = np.random.randint(0, 51, size=(5, 10))
M2 = np.random.randint(0, 51, size=(10, 5))


def Matrix_multip(M1, M2):
    rows = M1.shape[0]
    common_dimension = M1.shape[1]
    columns = M2.shape[1]

    M = np.zeros((rows, columns), dtype=int)

    for i in range(rows):
        for j in range(columns):
            for k in range(common_dimension):
                M[i, j] = M[i, j] + M1[i, k] * M2[k, j]

    return M


M = Matrix_multip(M1, M2)

#test
print("M1:")
print(M1)

print("\nM2:")
print(M2)

print("\nM1 * M2:")
print(M)

print("\nthe result is", np.array_equal(M, np.dot(M1, M2)))
```

### method2

```python
def Pascal_triangle(k):
    if not isinstance(k, int) or k < 1:
        print("error")
        return

    line = [1]

    for i in range(1, k):
        next_number = line[-1] * (k - i) // i
        line.append(next_number)

    return line

#test
print("The 100th line:")
print(Pascal_triangle(100))

print("\nThe 200th line:")
print(Pascal_triangle(200))
```

```python
import numpy as np

def Least_moves(x):
    if not isinstance(x, int) or x < 1:
        print("error")
        return

    def calculate(n):
        if n == 1:
            return 0

        if n % 2 == 0:
            return calculate(n // 2) + 1
        else:
            return calculate(n - 1) + 1

    result = calculate(x)
    print(result)
    return result

#test1
Least_moves(2)
Least_moves(5)

#test2
x = int(np.random.randint(1, 101))
print("x =", x)
print("Least_moves result:")
Least_moves(x)
```

## 5. Dynamic programming

```python
from itertools import product
import matplotlib.pyplot as plt

origin = "123456789"
operators = ["", "+", "-"]

# 穷举
all_expressions = []

for choices in product(operators, repeat=8):
    expression = origin[0]

    for i in range(8):
        expression = expression + choices[i] + origin[i + 1]

    all_expressions.append(expression)
# print(all_expressions)


def Find_expression(target):
    count = 0

    for expression in all_expressions:
        if eval(expression) == target:
            print(expression, "=", target)
            count = count + 1

    print("Total solutions:", count)
    return count


#5.1
Find_expression(50)


#5.2
Total_solutions = []

for target in range(1, 101):
    count = 0

    for expression in all_expressions:
        if eval(expression) == target:
            count = count + 1

    Total_solutions.append(count)


maximum = max(Total_solutions)
minimum = min(Total_solutions)

maximum_numbers = []
minimum_numbers = []

for number in range(1, 101):
    if Total_solutions[number - 1] == maximum:
        maximum_numbers.append(number)

    if Total_solutions[number - 1] == minimum:
        minimum_numbers.append(number)


print("maximum:", maximum)
print("numbers with maximum solutions:", maximum_numbers)

print("minimum:", minimum)
print("numbers with minimum solutions:", minimum_numbers)


plt.plot(range(1, 101), Total_solutions)
plt.xlabel("Target number")
plt.ylabel("Number of solutions")
plt.title("Total solutions from 1 to 100")
plt.grid()
plt.show()
```
